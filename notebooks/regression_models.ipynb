```python
import pandas as pd

flights = pd.read_csv("../data/cleaned/cleaned_flights.csv")
```

```python
flights["carrier_code"].value_counts()
```

```python
import statsmodels.formula.api as smf

model = smf.ols("departure_delay_minutes ~ departure_hour", data=flights).fit()

print(model.summary())
```

Departure hour was a statistically significant predictor of departure delay ($p < 0.001$), with delays increasing by 0.27 minutes for each additional hour later in the day. However, the model explained very little of the overall variation in delays ($R^2 = 0.001$), suggesting that departure hour alone is not a strong predictor of delay length.

```python
model2 = smf.ols("departure_delay_minutes ~ departure_hour + C(carrier_code) + C(day_of_week) + C(month) + C(origin_airport)",
data=flights).fit()

print(model2.summary())
```

```python
print(model2.params)
```

A multiple linear regression model was fit using departure delay by minutes as the response variable and departure hour, airline, day of week, month, and origin airport as predictors. American Airlines (AA), Friday, January, and LAX were used as the reference categories.

The model was statistically significant (p < 0.001), but the $R^2$ value remained low at 0.08, indicating that these variables explain only a small portion of the overall variation in departure delays.

Relative to American Airlines, Alaska, JetBlue, Delta, and United all had lower average departure delays after controlling for the other variables. Tuesday and Wednesday flights tended to have lower delays than Friday flights, while Sunday flights tended to have higher delays. Seasonal effects were also observed, with July and December showing higher average delays. Also, flights departing from SAN and SFO experienced slightly higher average delays than flights departing from LAX.

The results suggest that airline, departure time, day of week, season, and airport are all associated with departure delays, although additional factors likely account for most of the variation.

```python
logit_model = smf.logit("delayed_15 ~ departure_hour + C(carrier_code) + C(day_of_week) + C(month) + C(origin_airport)", data=flights).fit()

print(logit_model.summary())
```

```python
print(logit_model.params)
```

A logistic regression model was fit to predict whether a flight experienced a departure delay of at least 15 minutes. The predictors included departure hour, airline, day of week, month, and origin airport. American Airlines (AA), Friday, January, and LAX were used as the reference categories.

The model was statistically significant ($p < 0.001$) and produced an $R^2$ value of about 0.019, indicating that the included predictors explain a modest portion of the variation in whether a flight is delayed.
